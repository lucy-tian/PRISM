```python
import pandas as pd
import numpy as np
```

```python
traits = ['eGFR', 'FEV1_FVC_ratio', 'LDL-C', 'Lymphocyte_Count']
```

```python
binary_dfs = []
nonbinary_dfs = []
for t in traits:
    binary = pd.read_csv(f'binary_or_table/{t}_table.tsv', sep='\t')
    binary['trait'] = t
    non_binary = pd.read_csv(f'nonbinary_or_table/{t}_table.tsv', sep='\t')
    non_binary['trait'] = t
    binary_dfs.append(binary)
    nonbinary_dfs.append(non_binary)
```

```python
binary_df = pd.concat(binary_dfs)
```

```python
nonbinary_df = pd.concat(nonbinary_dfs)
```

```python
nonbinary_df['odds_ratio_per_SD'].mean()
```

```python
binary_df['odds_ratio'].mean()
```

```python
binary_df.to_csv('Binary_OR.tsv', sep='\t', index=False)
```

```python
nonbinary_df.to_csv('NonBinary_OR.tsv', sep='\t', index=False)
```
