```python
import pandas as pd
import numpy as np
```

```python
dir_p = '~/group/data/ENCODE4/cV2F/20231221/scores/mvp_afr/tissue/'
for t in ['BLOOD', 'KIDNEY', 'LIVER', 'LUNG']:
    dfs = []
    for i in np.arange(1, 23):
        df = pd.read_csv(dir_p + 'mvp_afr_all_annotation_chrmbpnet_' + t + '/mvp_afr_0.9_0.01_all_annotation_chrmbpnet_' + t + '_lava_ld_ld_maf.' + str(i) + '.cv2f.txt.gz', sep='\t')
        dfs.append(df)
    all_df = pd.concat(dfs)
    print(f'1% cutoff for {t}: {all_df["cV2F"].quantile(0.99)}')
```
